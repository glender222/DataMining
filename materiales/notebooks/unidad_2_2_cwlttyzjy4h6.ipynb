### Frecuencias Categóricas

```python
print('Frecuencia para Use Chip:')
print(df['Use Chip'].value_counts())
print('-'*10)
print('\nFrecuencia para Errors:')
print(df['Errors?'].value_counts())

print('-'*10)
print('\nFrecuencia para Is Fraud:')
print(df['Is Fraud?'].value_counts())
print('-'*10)

print('\nTop 10 Frecuencias para Merchant State:')
print(df['Merchant State'].value_counts().head(10))
```
