## 4.1 Numérica vs numérica


```python
var_num_1 = "Amount Received"
var_num_2 = "Amount Paid"

if var_num_1 in df.columns and var_num_2 in df.columns:
    display(df[[var_num_1, var_num_2]].describe())
    correlation = df[[var_num_1, var_num_2]].corr().iloc[0, 1]
    print(f"Correlación entre {var_num_1} y {var_num_2}: {correlation:.4f}")
else:
    print("Reemplaza los nombres por dos variables numéricas existentes.")
```

```python
if var_num_1 in df.columns and var_num_2 in df.columns:
    plt.figure(figsize=(7,5))
    plt.scatter(df[var_num_1], df[var_num_2], alpha=0.5)
    plt.xlabel(var_num_1)
    plt.ylabel(var_num_2)
    plt.title(f"{var_num_1} vs {var_num_2}")
    plt.show()
```

Se observa una asociación positiva entre `Amount Received` y `Amount Paid`: la correlación de Pearson es de **0,7396**, positiva y relativamente fuerte; en general, las transacciones con un monto pagado mayor tienden a recibir un monto mayor.
Aun así, la dispersión es alta y la escala del gráfico la dominan unos pocos montos extremos (el máximo llega a 3,6e12). Además, los montos pueden estar expresados en monedas distintas, por lo que la relación no implica que los valores coincidan ni demuestra causalidad.

```python
var_num_1 = "Amount Received"
var_num_2 = "Amount Paid"

if var_num_1 in df.columns and var_num_2 in df.columns:
    display(df[[var_num_1, var_num_2]].describe())
    correlation = df[[var_num_1, var_num_2]].corr().iloc[0, 1]
    print(f"Correlación entre {var_num_1} y {var_num_2}: {correlation:.4f}")
else:
    print("Reemplaza los nombres por dos variables numéricas existentes.")
```

```python
if var_num_1 in df.columns and var_num_2 in df.columns:
    plt.figure(figsize=(7,5))
    plt.scatter(df[var_num_1], df[var_num_2], alpha=0.5)
    plt.xlabel(var_num_1)
    plt.ylabel(var_num_2)
    plt.title(f"{var_num_1} vs {var_num_2}")
    plt.show()
```

El gráfico de dispersión confirma la misma relación: la mayoría de las transacciones se concentra en montos bajos y unas pocas operaciones con montos muy altos estiran la escala, de modo que la nube de puntos queda aplastada contra los ejes.
Por eso conviene revisar los montos extremos antes de interpretar el resultado y, si se quiere comparar montos de monedas distintas, analizarlos por separado o convertirlos a una divisa común. Recuerda que la correlación no implica causalidad.

#### 2. Amount Paid vs Is Laundering (monto pagado e indicador de lavado de dinero)
- **¿Qué relación observas?** La correlación calculada no muestra una asociación lineal apreciable entre el monto pagado y la etiqueta de lavado de dinero.
- **¿La correlación es positiva, negativa o cercana a cero?** Es **cercana a cero**. El resultado mostrado es **−0,0000**, lo que refleja el redondeo a cuatro decimales, no una relación negativa relevante.
- **¿La relación tiene sentido para el negocio?** Sí: el monto por sí solo no basta para identificar lavado de dinero. Este resultado no demuestra que el monto sea irrelevante ni descarta relaciones no lineales.
- **¿Podría existir una tercera variable explicando esa relación?** El formato de pago, la moneda o los patrones de transacciones de una cuenta podrían influir tanto en los montos como en su asociación con la etiqueta. Son posibilidades que habría que investigar.

#### 3. Amount Received vs Is Laundering (monto recibido e indicador de lavado de dinero)
- **¿Qué relación observas?** Tampoco se observa una asociación lineal apreciable entre el monto recibido y la etiqueta de lavado de dinero según la correlación calculada.
- **¿La correlación es positiva, negativa o cercana a cero?** Es **cercana a cero**, con un resultado mostrado de **−0,0000**.
- **¿La relación tiene sentido para el negocio?** Sí: recibir un monto elevado no significa necesariamente que exista lavado de dinero. Para investigar operaciones sospechosas se necesita más contexto que el monto recibido.
- **¿Podría existir una tercera variable explicando esa relación?** La moneda de recepción, el formato de pago o la frecuencia de operaciones de la cuenta podrían influir. El análisis realizado no permite confirmar esas explicaciones.

**Nota:** `Is Laundering` es una variable binaria (0/1), aunque esté almacenada como numérica. Su correlación de Pearson con los montos se puede calcular, pero no debe interpretarse como si fuera una medida continua. Además, comparar montos de distintas monedas sin convertirlos a una moneda común limita la interpretación.

> **Correlación no implica causalidad.** Una correlación cercana a cero no descarta otros tipos de asociación.


```python
print(df.select_dtypes(include="number").columns.tolist())
```

```python
pares = [
    ("Amount Received", "Amount Paid"),
    ("Amount Paid", "Is Laundering"),
    ("Amount Received", "Is Laundering")
]

for x, y in pares:
    if x not in df.columns or y not in df.columns:
        print(f"No se puede analizar {x} vs {y}: falta una columna.")
        continue

    datos = df[[x, y]].dropna()
    correlacion = datos[x].corr(datos[y])

    print(f"\n{x} vs {y}")
    print(f"Registros completos: {len(datos)}")
    print(f"Correlación de Pearson: {correlacion:.4f}")

    # Una muestra para que el gráfico no sea demasiado pesado.
    muestra = datos.sample(n=min(10000, len(datos)), random_state=42)

    plt.figure(figsize=(7, 4))
    plt.scatter(muestra[x], muestra[y], alpha=0.2, s=10)
    plt.xlabel(x)
    plt.ylabel(y)
    plt.title(f"{x} vs {y}")
    plt.tight_layout()
    plt.show()
```

```python
var_cat_1 = "Receiving Currency"
var_cat_2 = "Payment Currency"

if var_cat_1 in df.columns and var_cat_2 in df.columns:
    tabla = pd.crosstab(df[var_cat_1], df[var_cat_2])
    display(tabla)

    tabla_pct = pd.crosstab(
        df[var_cat_1],
        df[var_cat_2],
        normalize="index"
    ) * 100
    print("\nPorcentajes por fila:")
    display(tabla_pct.round(2))
else:
    print("Reemplaza los nombres por dos variables categóricas existentes.")
```

```python
var_cat_1 = "Payment Format"
var_cat_2 = "Payment Currency"

if var_cat_1 in df.columns and var_cat_2 in df.columns:
    # 1. Tabla de contingencia con frecuencias absolutas (conteo)
    tabla = pd.crosstab(df[var_cat_1], df[var_cat_2])
    display(tabla)

    # 2. Tabla de porcentajes normalizada por columna
    tabla_pct = (
        pd.crosstab(df[var_cat_1], df[var_cat_2], normalize="columns") * 100
    )
    print("\nPorcentajes por columna:")
    display(tabla_pct.round(2))
else:
    print("Reemplaza los nombres por dos variables categóricas existentes.")
```

```python
import matplotlib.pyplot as plt

# Usamos la tabla porcentual calculada anteriormente (transpuesta para graficar por divisa)
ax = tabla_pct.T.plot(
    kind="bar", stacked=True, figsize=(14, 7), colormap="tab10"
)

plt.title(
    "Distribución Porcentual de Métodos de Pago por Divisa",
    fontsize=14,
    pad=15,
)
plt.xlabel("Divisa (Payment Currency)", fontsize=12)
plt.ylabel("Porcentaje (%)", fontsize=12)
plt.legend(
    title="Payment Format",
    bbox_to_anchor=(1.02, 1),
    loc="upper left",
    frameon=True,
)
plt.xticks(rotation=45, ha="right")
plt.ylim(0, 100)
plt.grid(axis="y", linestyle="--", alpha=0.5)

plt.tight_layout()
plt.show()
```

# 6. Segmentación exploratoria

Aquí **segmentar** significa comparar grupos ya existentes en los datos.
No entrenaremos clustering.


```python
variable_segmento = "Payment Format"

if variable_segmento in df.columns:
    print("Cantidad por segmento:")
    display(df[variable_segmento].value_counts(dropna=False))

    print("\nPorcentaje por segmento:")
    display((df[variable_segmento].value_counts(normalize=True, dropna=False) * 100).round(2))
else:
    print("Reemplaza 'REEMPLAZAR' por una variable categórica.")
```

```python
variables_perfil = ["Amount Received", "Amount Paid"]
variables_validas = [v for v in variables_perfil if v in df.columns]

if variable_segmento in df.columns and variables_validas:
    perfil = df.groupby(variable_segmento)[variables_validas].mean()
    display(perfil.round(2))
else:
    print("Reemplaza por variables existentes.")
```

Existen marcadas diferencias entre los métodos de pago según su volumen y capital promedio: mientras Cheque y Tarjeta de Crédito dominan en cantidad de transacciones, los canales interbancarios como ACH, Cheque y Wire concentran los montos económicos promedio más elevados.

# 7. Representatividad

Un dataset puede tener buena calidad técnica y aun así **no representar correctamente a la población objetivo**.


```python
variable_representatividad = "Payment Format"

if variable_representatividad in df.columns:
    representatividad = (
        df[variable_representatividad]
        .value_counts(normalize=True, dropna=False)
        .mul(100)
        .round(2)
    )
    display(representatividad)
else:
    print("Reemplaza por una variable como ciudad, región, género, segmento, etc.")
```

La distribución de métodos de pago no es uniforme, existiendo una fuerte concentración en cheques y tarjetas frente a canales minoritarios como Wire y Bitcoin, lo que podría restar representatividad a estos últimos en el comportamiento general del dataset.

```python
# Hallazgos calculados sobre LI-Small_Trans.csv; no modifica df.
columnas_requeridas = [
    'Amount Received', 'Amount Paid', 'Receiving Currency',
    'Payment Currency', 'Payment Format', 'Is Laundering'
]
faltantes = [col for col in columnas_requeridas if col not in df.columns]
if faltantes:
    raise ValueError(f'Faltan columnas para documentar los hallazgos: {faltantes}')

hallazgos = []

# 1. Correlación global: mezcla monedas, por lo que requiere cautela.
montos = df[['Amount Received', 'Amount Paid']].dropna()
r = montos['Amount Received'].corr(montos['Amount Paid'])
hallazgos.append({
    'Hallazgo': 'Relación lineal entre monto recibido y monto pagado',
    'Variables_involucradas': 'Amount Received - Amount Paid',
    'Evidencia': f'{len(montos):,} pares completos; correlación de Pearson = {r:.4f}.',
    'Interpretacion': 'La correlación resume la asociación lineal observada, no causalidad. Al mezclar monedas, su interpretación económica es limitada; conviene analizar cada combinación de monedas.'
})

# 2. Coincidencia de las monedas dentro de cada transacción.
monedas = df[['Receiving Currency', 'Payment Currency']].dropna()
coinciden = monedas['Receiving Currency'].eq(monedas['Payment Currency'])
pct_misma = coinciden.mean() * 100
hallazgos.append({
    'Hallazgo': 'Coincidencia entre moneda de recepción y moneda de pago',
    'Variables_involucradas': 'Receiving Currency - Payment Currency',
    'Evidencia': f'{len(monedas):,} pares completos; {int(coinciden.sum()):,} ({pct_misma:.2f} %) tienen la misma moneda.',
    'Interpretacion': 'La proporción describe la coincidencia de monedas. No prueba independencia ni causalidad; las operaciones con monedas distintas requieren considerar tipos de cambio.'
})

# 3. Tamaños de los grupos por formato de pago.
formatos = df['Payment Format'].value_counts()
if not formatos.empty:
    mayor, menor = formatos.index[0], formatos.index[-1]
    evidencia_formatos = (
        f'Mayor frecuencia: {mayor}, {int(formatos.iloc[0]):,} '
        f'({formatos.iloc[0] / formatos.sum() * 100:.2f} %); '
        f'menor frecuencia: {menor}, {int(formatos.iloc[-1]):,} '
        f'({formatos.iloc[-1] / formatos.sum() * 100:.2f} %). '
        f'Faltantes: {int(df["Payment Format"].isna().sum()):,}.'
    )
else:
    evidencia_formatos = 'No hay formatos de pago no nulos para comparar.'
hallazgos.append({
    'Hallazgo': 'Concentración de registros por formato de pago',
    'Variables_involucradas': 'Payment Format',
    'Evidencia': evidencia_formatos,
    'Interpretacion': 'Los tamaños de grupo deben acompañar las comparaciones. Una frecuencia baja no demuestra subrepresentación sin una referencia de la población objetivo.'
})

# 4. Distribución de la etiqueta, sin confundirla con un dictamen legal.
etiquetas = df['Is Laundering'].value_counts(dropna=False)
evidencia_etiquetas = '; '.join(
    f'{etiqueta}: {int(n):,} ({n / len(df) * 100:.2f} %)'
    for etiqueta, n in etiquetas.items()
)
hallazgos.append({
    'Hallazgo': 'Distribución de la etiqueta de lavado',
    'Variables_involucradas': 'Is Laundering',
    'Evidencia': evidencia_etiquetas or 'No hay registros.',
    'Interpretacion': 'Estos conteos permiten evaluar el desbalance de clases. Las comparaciones con grupos pequeños requieren cautela; la etiqueta no constituye una confirmación legal de lavado.'
})

# 5. Comparación de medianas dentro de la moneda de pago más frecuente.
monedas_pago = df['Payment Currency'].value_counts()
if not monedas_pago.empty:
    moneda = monedas_pago.index[0]
    resumen_moneda = (
        df.loc[df['Payment Currency'].eq(moneda)]
        .groupby('Payment Format')['Amount Paid']
        .agg(cantidad='count', mediana='median')
        .dropna(subset=['mediana'])
        .sort_values('mediana', ascending=False)
    )
    evidencia_mediana = f'Moneda analizada: {moneda}. ' + '; '.join(
        f'{formato}: mediana={fila["mediana"]:,.2f}, n={int(fila["cantidad"]):,}'
        for formato, fila in resumen_moneda.iterrows()
    )
else:
    evidencia_mediana = 'No hay monedas de pago no nulas para comparar.'
hallazgos.append({
    'Hallazgo': 'Comparación del monto pagado por formato en una misma moneda',
    'Variables_involucradas': 'Amount Paid - Payment Format - Payment Currency',
    'Evidencia': evidencia_mediana,
    'Interpretacion': 'Las medianas permiten comparar montos sin mezclar monedas. Las diferencias son descriptivas, deben leerse junto con los tamaños de grupo y no implican que el formato cause un monto determinado. El análisis se limita a la moneda seleccionada.'
})

hallazgos_relaciones = pd.DataFrame(hallazgos)
display(hallazgos_relaciones)
```

# 11. Conclusión final de la Unidad 2

**¿Qué aprendimos sobre nuestros datos y qué debemos considerar antes de pasar a Preparación de Datos?**

Aprendimos que `LI-Small_Trans.csv` contiene **transacciones financieras**. Cada fila representa una operación, con variables como monto pagado, monto recibido, monedas, formato de pago y la etiqueta `Is Laundering`. En el análisis de los tres pares numéricos se reportaron **6.924.049 registros completos por par**; este dato no demuestra por sí solo que todas las columnas estén completas.

- **Calidad de los datos:** antes de prepararlos debemos comprobar los valores faltantes, las filas repetidas, los tipos de datos, las categorías y la validez de montos y fechas. Una fila repetida no debe eliminarse automáticamente, porque podría corresponder a una operación legítima. Los montos extremos también requieren revisión: no necesariamente son errores.

- **Relaciones relevantes:** el monto recibido y el monto pagado presentan una correlación de Pearson de **0,7396**, positiva y relativamente fuerte, según los resultados documentados. En cambio, las correlaciones de ambos montos con `Is Laundering` son cercanas a cero. Esto no significa que los montos sean irrelevantes ni que no existan relaciones no lineales. También se documentó que **el 99,98 % de las operaciones con formato Bitcoin utiliza Bitcoin como moneda de pago**. Estas asociaciones no demuestran causalidad.

- **Representatividad:** las categorías tienen tamaños distintos y las proporciones documentadas muestran una concentración importante en dólares estadounidenses y euros. Sin embargo, tener muchos registros no garantiza que el conjunto represente todas las transacciones financieras. Debemos contrastar su cobertura temporal, las monedas, los formatos de pago y el origen de los datos con la población que deseamos estudiar.

- **Sesgos:** los grupos más frecuentes pueden dominar los resultados globales y ocultar el comportamiento de los grupos pequeños. También debemos verificar el desbalance de `Is Laundering`, ya que una clase minoritaria dificultaría las comparaciones y una evaluación posterior basada únicamente en exactitud podría resultar engañosa.

- **Limitaciones:** los montos expresados en diferentes monedas no son directamente comparables. Los valores extremos pueden influir en promedios y correlaciones, y las gráficas basadas en muestras pueden omitir operaciones poco frecuentes. La etiqueta `Is Laundering` debe interpretarse según su definición y procedencia; no constituye por sí sola una confirmación legal de lavado. Sin información suficiente sobre cómo se generó o seleccionó el dataset, no podemos garantizar su generalización al sistema financiero real.

**Antes de pasar a Preparación de Datos**, debemos definir reglas justificadas para tratar nulos, duplicados y valores inválidos; validar tipos y categorías; decidir si analizaremos los montos por moneda o si los convertiremos con tipos de cambio adecuados; y documentar el tratamiento de valores extremos. Si posteriormente se construyen modelos, habrá que considerar el desbalance de clases y una partición que evite fugas de información entre operaciones relacionadas o periodos. **No se deben eliminar ni transformar datos sin evidencia y justificación.**
